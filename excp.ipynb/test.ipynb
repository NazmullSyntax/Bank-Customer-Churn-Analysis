```python
import pandas as pd
import numpy as np
from pathlib import Path

np.random.seed(42)
n = 10000

data = {
    'CustomerId': np.arange(10000000, 10000000 + n),
    'Surname': [f'Customer_{i}' for i in range(n)],
    'CreditScore': np.random.randint(350, 850, n),
    'Geography': np.random.choice(['France', 'Germany', 'Spain'], n, p=[0.5, 0.25, 0.25]),
    'Gender': np.random.choice(['Male', 'Female'], n, p=[0.55, 0.45]),
    'Age': np.random.randint(18, 92, n),
    'Tenure': np.random.randint(0, 11, n),
    'Balance': np.round(np.random.uniform(0, 250000, n), 2),
    'NumOfProducts': np.random.choice([1, 2, 3, 4], n, p=[0.5, 0.45, 0.04, 0.01]),
    'HasCrCard': np.random.choice([0, 1], n, p=[0.3, 0.7]),
    'IsActiveMember': np.random.choice([0, 1], n, p=[0.5, 0.5]),
    'EstimatedSalary': np.round(np.random.uniform(11, 200000, n), 2),
}

df = pd.DataFrame(data)

# Create realistic churn based on features
churn_prob = (
    0.05
    + 0.15 * (df['Age'] > 50)
    + 0.20 * (df['IsActiveMember'] == 0)
    + 0.15 * (df['NumOfProducts'] > 2)
    + 0.10 * (df['Geography'] == 'Germany')
    + 0.05 * (df['Balance'] > 100000)
)
churn_prob = churn_prob.clip(0, 1)
df['Exited'] = np.random.binomial(1, churn_prob)

Path("data").mkdir(parents=True, exist_ok=True)
df.to_csv("data/churn.csv", index=False)
print(f"Dataset saved: {df.shape}")
print(df.head())
```
