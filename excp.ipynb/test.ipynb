```python
import pandas as pd
import numpy as np
from pathlib import Path

np.random.seed(42)
n = 10000

data = {
    'CustomerId': np.arange(10000000, 10000000 + n),
    'Surname': [f'Customer_{i}' for i in range(n)],
    'CreditScore': np.random.randint(350, 850, n),
    'Geography': np.random.choice(['France', 'Germany', 'Spain'], n, p=[0.5, 0.25, 0.25]),
    'Gender': np.random.choice(['Male', 'Female'], n, p=[0.55, 0.45]),
    'Age': np.random.randint(18, 92, n),
    'Tenure': np.random.randint(0, 11, n),
    'Balance': np.round(np.random.uniform(0, 250000, n), 2),
    'NumOfProducts': np.random.choice([1, 2, 3, 4], n, p=[0.5, 0.45, 0.04, 0.01]),
    'HasCrCard': np.random.choice([0, 1], n, p=[0.3, 0.7]),
    'IsActiveMember': np.random.choice([0, 1], n, p=[0.5, 0.5]),
    'EstimatedSalary': np.round(np.random.uniform(11, 200000, n), 2),
}

df = pd.DataFrame(data)

# Create realistic churn based on features
churn_prob = (
    0.05
    + 0.15 * (df['Age'] > 50)
    + 0.20 * (df['IsActiveMember'] == 0)
    + 0.15 * (df['NumOfProducts'] > 2)
    + 0.10 * (df['Geography'] == 'Germany')
    + 0.05 * (df['Balance'] > 100000)
)
churn_prob = churn_prob.clip(0, 1)
df['Exited'] = np.random.binomial(1, churn_prob)

Path("data").mkdir(parents=True, exist_ok=True)
df.to_csv("data/churn.csv", index=False)
print(f"Dataset saved: {df.shape}")
print(df.head())
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder

def engineer_features(df):
    df = df.copy()
    
    # Drop non-informative columns
    df = df.drop(columns=['CustomerId', 'Surname'], errors='ignore')
    
    # Feature engineering
    df['Balance_to_Salary'] = df['Balance'] / (df['EstimatedSalary'] + 1)
    df['Age_Tenure_Ratio'] = df['Age'] / (df['Tenure'] + 1)
    df['CreditScore_Age'] = df['CreditScore'] * df['Age'] / 1000
    df['Is_Senior'] = (df['Age'] >= 60).astype(int)
    df['Zero_Balance'] = (df['Balance'] == 0).astype(int)
    df['Products_Per_Tenure'] = df['NumOfProducts'] / (df['Tenure'] + 1)
    
    return df

def prepare_data(df, target='Exited', test_size=0.2, random_state=42):
    df = engineer_features(df)
    
    # Encode categoricals
    le_geo = LabelEncoder()
    le_gen = LabelEncoder()
    df['Geography'] = le_geo.fit_transform(df['Geography'])
    df['Gender'] = le_gen.fit_transform(df['Gender'])
    
    X = df.drop(columns=[target])
    y = df[target]
    
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state, stratify=y
    )
    
    scaler = StandardScaler()
    num_cols = X_train.select_dtypes(include=['number']).columns
    X_train[num_cols] = scaler.fit_transform(X_train[num_cols])
    X_test[num_cols] = scaler.transform(X_test[num_cols])
    
    return X_train, X_test, y_train, y_test, scaler, le_geo, le_gen, X.columns.tolist()
```

```python
import pandas as pd
import numpy as np
import joblib
import os
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                              f1_score, roc_auc_score, classification_report,
                              confusion_matrix, roc_curve)
%pip install -q imbalanced-learn

from imblearn.over_sampling import SMOTE
%pip install -q xgboost
from xgboost import XGBClassifier
import matplotlib.pyplot as plt
import seaborn as sns

def load_data():
    return df.copy()
# prepare_data is already defined in the earlier notebook cell

os.makedirs('models', exist_ok=True)
os.makedirs('reports/figures', exist_ok=True)

def evaluate_model(name, model, X_test, y_test):
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]
    return {
        'Model': name,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1': f1_score(y_test, y_pred),
        'ROC_AUC': roc_auc_score(y_test, y_prob)
    }, y_prob

def train_models():
    df = load_data()
    X_train, X_test, y_train, y_test, scaler, le_geo, le_gen, features = prepare_data(df)
    
    # Handle class imbalance
    smote = SMOTE(random_state=42)
    X_train_sm, y_train_sm = smote.fit_resample(X_train, y_train)
    print(f"Original: {y_train.value_counts().to_dict()}")
    print(f"After SMOTE: {pd.Series(y_train_sm).value_counts().to_dict()}")
    
    models = {
        'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
        'Random Forest': RandomForestClassifier(n_estimators=200, max_depth=10,
                                                 random_state=42, n_jobs=-1),
        'Gradient Boosting': GradientBoostingClassifier(n_estimators=200,
                                                         random_state=42),
        'XGBoost': XGBClassifier(n_estimators=200, max_depth=6, learning_rate=0.1,
                                  random_state=42, eval_metric='logloss')
    }
    
    results = []
    probs = {}
    for name, model in models.items():
        print(f"\nTraining {name}...")
        model.fit(X_train_sm, y_train_sm)
        metrics, y_prob = evaluate_model(name, model, X_test, y_test)
        results.append(metrics)
        probs[name] = y_prob
        print(f"  Accuracy: {metrics['Accuracy']:.4f} | F1: {metrics['F1']:.4f} | ROC-AUC: {metrics['ROC_AUC']:.4f}")
    
    results_df = pd.DataFrame(results).sort_values('ROC_AUC', ascending=False)
    print("\n" + "=" * 60)
    print("MODEL COMPARISON")
    print("=" * 60)
    print(results_df.to_string(index=False))
    
    results_df.to_csv('reports/model_comparison.csv', index=False)
    
    # Best model
    best_name = results_df.iloc[0]['Model']
    best_model = models[best_name]
    print(f"\n🏆 Best Model: {best_name}")
    
    # Full evaluation of best model
    y_pred = best_model.predict(X_test)
    print("\n" + classification_report(y_test, y_pred, target_names=['Retained', 'Churned']))
    
    # Confusion matrix
    plt.figure(figsize=(6, 5))
    sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Blues',
                xticklabels=['Retained', 'Churned'], yticklabels=['Retained', 'Churned'])
    plt.title(f'Confusion Matrix - {best_name}')
    plt.ylabel('Actual')
    plt.xlabel('Predicted')
    plt.tight_layout()
    plt.savefig('reports/figures/confusion_matrix.png', dpi=100)
    plt.close()
    
    # ROC Curves
    plt.figure(figsize=(8, 6))
    for name, y_prob in probs.items():
        fpr, tpr, _ = roc_curve(y_test, y_prob)
        auc = roc_auc_score(y_test, y_prob)
        plt.plot(fpr, tpr, label=f'{name} (AUC={auc:.3f})')
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('ROC Curves')
    plt.legend()
    plt.tight_layout()
    plt.savefig('reports/figures/roc_curves.png', dpi=100)
    plt.close()
    
    # Feature importance (if tree-based)
    if hasattr(best_model, 'feature_importances_'):
        imp = pd.DataFrame({'Feature': features,
                            'Importance': best_model.feature_importances_}
                           ).sort_values('Importance', ascending=False)
        plt.figure(figsize=(10, 8))
        sns.barplot(data=imp.head(15), x='Importance', y='Feature', palette='viridis')
        plt.title(f'Top 15 Feature Importances - {best_name}')
        plt.tight_layout()
        plt.savefig('reports/figures/feature_importance.png', dpi=100)
        plt.close()
        print("\nTop 10 Features:")
        print(imp.head(10).to_string(index=False))
    
    # Save artifacts
    joblib.dump({
        'model': best_model,
        'scaler': scaler,
        'le_geo': le_geo,
        'le_gen': le_gen,
        'features': features,
        'model_name': best_name
    }, 'models/churn_model.pkl')
    print("\n✅ Model saved to models/churn_model.pkl")
    
    return best_model, results_df

if __name__ == '__main__':
    train_models()
```

```python
import joblib
import pandas as pd
import numpy as np

def load_artifacts(path='models/churn_model.pkl'):
    return joblib.load(path)

def predict_single(customer: dict, artifacts=None):
    if artifacts is None:
        artifacts = load_artifacts()
    
    df = pd.DataFrame([customer])
    
    # Feature engineering (must match training)
    df['Balance_to_Salary'] = df['Balance'] / (df['EstimatedSalary'] + 1)
    df['Age_Tenure_Ratio'] = df['Age'] / (df['Tenure'] + 1)
    df['CreditScore_Age'] = df['CreditScore'] * df['Age'] / 1000
    df['Is_Senior'] = (df['Age'] >= 60).astype(int)
    df['Zero_Balance'] = (df['Balance'] == 0).astype(int)
    df['Products_Per_Tenure'] = df['NumOfProducts'] / (df['Tenure'] + 1)
    
    df['Geography'] = artifacts['le_geo'].transform(df['Geography'])
    df['Gender'] = artifacts['le_gen'].transform(df['Gender'])
    
    df = df[artifacts['features']]
    df[artifacts['features']] = artifacts['scaler'].transform(df)
    
    prob = artifacts['model'].predict_proba(df)[0, 1]
    pred = int(prob >= 0.5)
    
    return {
        'churn_probability': round(float(prob), 4),
        'churn_prediction': 'Churned' if pred else 'Retained',
        'risk_level': 'High' if prob > 0.7 else 'Medium' if prob > 0.4 else 'Low'
    }

if __name__ == '__main__':
    customer = {
        'CreditScore': 600,
        'Geography': 'Germany',
        'Gender': 'Female',
        'Age': 55,
        'Tenure': 3,
        'Balance': 120000,
        'NumOfProducts': 3,
        'HasCrCard': 1,
        'IsActiveMember': 0,
        'EstimatedSalary': 80000
    }
    result = predict_single(customer)
    print("\nPrediction Result:")
    for k, v in result.items():
        print(f"  {k}: {v}")
```
